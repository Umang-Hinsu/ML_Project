# 🏆 Task 5: Master Classification Model Comparison
## Project: CrediGuard AI - Loan Default Prediction

This notebook loads the exported metrics from all 5 trained models and presents a unified summary table and comparison chart.

## 1. Import Libraries & Load Model JSON Results

```
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

json_files = [
    'result_LogisticRegressionModel.json',
    'result_DecisionTreeModel.json',
    'result_SVCModel.json',
    'result_KNNModel.json',
    'result_NaiveBaseModel.json'
]

results = []
for filename in json_files:
    if os.path.exists(filename):
        with open(filename, 'r', encoding='utf-8') as f:
            results.append(json.load(f))

df_results = pd.DataFrame(results)
print("Loaded Results Count:", len(df_results))
```

## 2. Display Task 5 Master Comparison Table

```
columns_to_show = ['model', 'accuracy', 'precision', 'recall', 'f1_score', 'cv_mean_f1', 'cv_std_f1', 'fit_status']
df_display = df_results[columns_to_show].sort_values(by='f1_score', ascending=False)

print("=" * 95)
print(" 📊 TASK 5 CLASSIFICATION MODEL COMPARISON TABLE")
print("=" * 95)
print(df_display.to_string(index=False))
```

## 3. Plot F1-Score Bar Chart Comparison

```
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=df_display, x='model', y='f1_score', palette='magma')
plt.title('Task 5 Classification Model Comparison - F1-Score', fontsize=14, fontweight='bold')
plt.ylabel('F1-Score', fontsize=12)
plt.xlabel('Classification Model', fontsize=12)
plt.ylim(0, 1.0)
plt.xticks(rotation=15)

for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(f"{height:.4f}",
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', fontsize=11, fontweight='bold', xytext=(0, 3),
                    textcoords='offset points')

plt.tight_layout()
plt.show()
```

## 4. Export Master Table to Excel File

```
excel_filename = 'Task5_Classification_Models_Summary.xlsx'
with pd.ExcelWriter(excel_filename, engine='openpyxl') as writer:
    df_display.to_excel(writer, sheet_name='Task 5 Model Comparison', index=False)

print(f"Master comparison table saved to Excel file: {excel_filename}")
```
