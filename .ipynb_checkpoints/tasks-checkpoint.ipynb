### Data cleaning and pre processing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv("Loan_default.csv")
```

```python
df.drop("LoanID", axis=1, inplace=True)
```

```python
df.dtypes
```

### find categorical columns

```python
df.select_dtypes(include="object").columns
```

### Encode Categorical Columns

```python
## for education cols
df["Education"] = df["Education"].map({
    "High School": 0,
    "Bachelor's": 1,
    "Master's": 2,
    "PhD": 3
})
```

```python
## for Employment Type
df["EmploymentType"] = df["EmploymentType"].map({
    "Unemployed": 0,
    "Part-time": 1,
    "Full-time": 2,
    "Self-employed": 3
})
```

```python
## for Marital Status
df["MaritalStatus"] = df["MaritalStatus"].map({
    "Single": 0,
    "Married": 1,
    "Divorced": 2
})
```

```python
## for Has Mortgage
df["HasMortgage"] = df["HasMortgage"].map({
    "No": 0,
    "Yes": 1
})
```

```python
## for Has Dependents
df["HasDependents"] = df["HasDependents"].map({
    "No": 0,
    "Yes": 1
})
```

```python
## for Loan Purpose
df["LoanPurpose"] = df["LoanPurpose"].map({
    "Auto": 0,
    "Business": 1,
    "Education": 2,
    "Home": 3,
    "Other": 4
})
```

```python
## for Has Co-Signer
df["HasCoSigner"] = df["HasCoSigner"].map({
    "No": 0,
    "Yes": 1
})
```

```python
df
```

### to convert in range between 0 to 1

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[['Age','Income','CreditScore','LoanAmount']] = scaler.fit_transform(df[['Age','Income','CreditScore','LoanAmount']])
```

# Task-3

### Part A (Scikit-learn Linear Regression)

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Load Dataset
df = pd.read_csv("Loan_default.csv")

# Feature and Target
x = df[['LoanAmount']]
y = df['Default']

# Train Model
model = LinearRegression()
model.fit(x, y)

# Print Parameters
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)

# Predict
new_customer = pd.DataFrame({
    'LoanAmount': [200000]
})

prediction = model.predict(new_customer)

print("Prediction Value:", prediction[0])

if prediction[0] >= 0.5:
    print("Predicted Default: 1")
else:
    print("Predicted Default: 0")
```

### Part B (Gradient Descent from Scratch)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Loan_default.csv")

X = df['LoanAmount'].values.astype(float)
Y = df['Default'].values.astype(float)

n = len(X)

m, c = 0.0, 0.0
lr = 0.01
epochs = 10

loss_history = []

for i in range(epochs):

    Y_pred = m * X + c

    error = Y_pred - Y

    loss = (1/n) * np.sum(error ** 2)

    loss_history.append(loss)

    dm = (2/n) * np.sum(error * X)

    dc = (2/n) * np.sum(error)

    m = m - lr * dm

    c = c - lr * dc

print("Final Parameters")
print("m =", m)
print("c =", c)

# plt.plot(loss_history)
# plt.xlabel("Epoch")
# plt.ylabel("Loss")
# plt.title("Loss vs Epoch")
# plt.show()
```
