# 📊 Task 5: Decision Tree Model (Full Tuned Pipeline)
## Project: CrediGuard AI - Loan Default Prediction (Classification)

This notebook trains a **Decision Tree** model using the full Task 5 pipeline:
1. **Model Evaluation Metrics** (Accuracy, Precision, Recall, F1-Score)
2. **Overfitting / Underfitting Check** (Train F1 vs Test F1 Score comparison)
3. **5-Fold Cross-Validation** (Average F1 score & Spread / Std Dev)
4. **Hyperparameter Tuning** using `GridSearchCV`

### Step 1: Import Required Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import joblib
import warnings

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

warnings.filterwarnings('ignore')
```

### Step 2: Load and Clean Dataset

```
# Load data from CSV file
df = pd.read_csv('Loan_default.csv')

# Remove LoanID column as it is just an identifier
if 'LoanID' in df.columns:
    df = df.drop(columns=['LoanID'])

# Drop duplicate rows
df = df.drop_duplicates()

print("Dataset Shape:", df.shape)
```

### Step 3: Encode Categorical Features

```
# Binary columns (Yes/No -> 1/0)
binary_cols = ['HasMortgage', 'HasDependents', 'HasCoSigner']
for col in binary_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])

# Multi-class text columns (One-Hot Encoding)
categorical_cols = ['Education', 'EmploymentType', 'MaritalStatus', 'LoanPurpose']
ohe = OneHotEncoder(sparse_output=False)
encoded_features = ohe.fit_transform(df[categorical_cols])
encoded_df = pd.DataFrame(encoded_features, columns=ohe.get_feature_names_out(categorical_cols))

# Combine encoded columns back with dataset
df = pd.concat([df.drop(columns=categorical_cols), encoded_df], axis=1)
```

### Step 4: Split Features (X) and Target (y)

```
X = df.drop(columns=['Default'])
y = df['Default']

# Scale feature values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split 80% Train and 20% Test
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
```

### Step 5: Baseline Model Training & Initial Metrics

```
# Fit initial Decision Tree model
baseline_model = DecisionTreeClassifier(max_depth=10, random_state=42)
baseline_model.fit(X_train, y_train)

# Make predictions
train_preds = baseline_model.predict(X_train)
test_preds = baseline_model.predict(X_test)

# Calculate Scores
train_f1 = f1_score(y_train, train_preds)
test_acc = accuracy_score(y_test, test_preds)
test_prec = precision_score(y_test, test_preds, zero_division=0)
test_rec = recall_score(y_test, test_preds, zero_division=0)
test_f1 = f1_score(y_test, test_preds, zero_division=0)

print("=== Baseline Decision Tree Performance ===")
print(f"Train F1-Score: {train_f1:.4f}")
print(f"Test Accuracy : {test_acc:.4f}")
print(f"Test Precision: {test_prec:.4f}")
print(f"Test Recall   : {test_rec:.4f}")
print(f"Test F1-Score : {test_f1:.4f}")
```

### Step 6: Check Overfitting / Underfitting

```
# Diagnostic check: Compare Train F1 vs Test F1 score
score_diff = train_f1 - test_f1

if score_diff > 0.08:
    fit_status = "Overfitting [Warning] (Train score much higher than Test score)"
elif train_f1 < 0.20 and test_f1 < 0.20:
    fit_status = "Underfitting [Warning] (Both Train and Test scores are low)"
else:
    fit_status = "Good Fit [Good Fit] (Train and Test scores are close)"

print("Fit Diagnostic:", fit_status)
```

### Step 7: 5-Fold Cross-Validation

```
# Run 5-Fold Stratified Cross Validation
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(baseline_model, X_train[:15000], y_train[:15000], cv=skf, scoring='f1', n_jobs=1)

cv_mean = np.mean(cv_scores)
cv_std = np.std(cv_scores)

print("=== 5-Fold Cross-Validation Results ===")
print("Fold Scores:", [round(score, 4) for score in cv_scores])
print(f"Average CV F1-Score: {cv_mean:.4f}")
print(f"Score Spread (Std): {cv_std:.4f}")
```

### Step 8: Hyperparameter Tuning using GridSearchCV

```
# Grid search parameters for Decision Tree
param_grid = {
    'max_depth': [6, 10, 15],
    'min_samples_split': [10, 20, 50],
    'criterion': ['gini', 'entropy']
}

grid_search = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=param_grid,
    cv=3,
    scoring='f1',
    n_jobs=1
)

grid_search.fit(X_train[:20000], y_train[:20000])
best_params = grid_search.best_params_

print("Best Parameters Found:", best_params)

# Train final best model on full training set
best_model = grid_search.best_estimator_
best_model.fit(X_train, y_train)

# Re-test tuned model
final_preds = best_model.predict(X_test)
final_f1 = f1_score(y_test, final_preds, zero_division=0)
final_acc = accuracy_score(y_test, final_preds)
final_prec = precision_score(y_test, final_preds, zero_division=0)
final_rec = recall_score(y_test, final_preds, zero_division=0)

print(f"Retested Tuned F1-Score: {final_f1:.4f}")
```

### Step 9: Plot Confusion Matrix & Save Artifacts

```
# Plot Confusion Matrix
cm = confusion_matrix(y_test, final_preds)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Default (0)', 'Default (1)'],
            yticklabels=['No Default (0)', 'Default (1)'])
plt.title('Confusion Matrix - Tuned Decision Tree')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# Save trained model file
joblib.dump(best_model, 'DecisionTreeModel.pkl')

# Export JSON summary
results = {
    "model": "Decision Tree",
    "accuracy": round(final_acc, 4),
    "precision": round(final_prec, 4),
    "recall": round(final_rec, 4),
    "f1_score": round(final_f1, 4),
    "cv_mean_f1": round(cv_mean, 4),
    "cv_std_f1": round(cv_std, 4),
    "fit_status": fit_status,
    "best_params": str(best_params)
}

with open('result_DecisionTreeModel.json', 'w', encoding='utf-8') as f:
    json.dump(results, f, indent=2)

print("Saved results to result_DecisionTreeModel.json")
```
