## Outlier Detection

```python
outlier_columns = [
    "Age",
    "Income",
    "LoanAmount",
    "CreditScore",
    "MonthsEmployed",
    "NumCreditLines",
    "InterestRate",
    "LoanTerm",
    "DTIRatio"
]
```

```python
for column in outlier_columns:

    plt.figure(figsize=(8, 4))

    plt.boxplot(df_clean[column], vert=False)

    plt.xlabel(column)
    plt.title("Boxplot of " + column)

    plt.show()
```

```python
df_clean[outlier_columns].hist(
    figsize=(15, 12),
    bins=20
)

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

Observation:

Histograms were used to examine the distribution of numerical features.
The features are distributed across their respective ranges without obvious
invalid or highly extreme values.

## Detecting Outliers Using IQR Method

```python
for column in outlier_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ]

    print(column, ":", len(outliers))
```
